# Polynomial Regression with Gradient Descent

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_data(num_points=100):
    # Generate synthetic data for polynomial regression
    np.random.seed(42)
    X = 2 * np.random.rand(num_points, 1)
    y = 3 + 4 * X + np.random.randn(num_points, 1)
    return X, y

def plot_regression_line(X, y, theta):
    # Plot the regression line
    plt.scatter(X, y, color='blue')
    x_range = np.linspace(min(X), max(X), 100).reshape(-1, 1)
    x_range_poly = polynomial_features(x_range, degree)
    y_range_pred = x_range_poly.dot(theta)
    plt.plot(x_range, y_range_pred, color='red')
    plt.xlabel('X')
    plt.ylabel('y')
    plt.title('Polynomial Regression')
    plt.show()
```

# Input Features given degree

```python
def polynomial_features(X, degree):
    # Generate polynomial features up to the specified degree
    X_poly = np.ones_like(X)  # Initialize with X^0 (bias term)
    X = np.array(X)

    for i in range(1, degree + 1):
        # column-wise concatenate : np.concatenate(a,b)
        # todo 1
        raised = X**i

        X_poly = np.concatenate((X_poly, raised), axis=1)

    return X_poly

data = generate_data()
print(data)
poly_data = polynomial_features(X=data, degree=2)
```
